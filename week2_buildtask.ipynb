```python
class StockPortfolio:
    def __init__(self, owner_name):
        self.owner = owner_name
        self.holdings = {}  # {stock: {quantity, buy_price}}

    def buy(self, stock_name, buy_price, quantity):
        self.holdings[stock_name] = {
            'quantity': quantity,
            'buy_price': buy_price
        }
        total = buy_price * quantity
        print(f"✅ Bought {quantity} shares of{stock_name} @ ₹{buy_price} = ₹{total}")

    def portfolio_value(self, current_prices):
        total = 0
        for stock, details in self.holdings.items():
            current = current_prices.get(stock, details['buy_price'])
            value = current * details['quantity']
            total += value
        return total

    def pnl(self, current_prices):
        invested = 0
        for d in self.holdings.values():
          invested += d['buy_price'] * d['quantity']

        current_value = self.portfolio_value(current_prices)
        profit = current_value - invested
        pct = (profit / invested) * 100
        return profit, pct

    def show(self, current_prices):
        print(f"\n📊{self.owner}'s Portfolio")
        print("=" * 40)
        for stock, details in self.holdings.items():
            buy_val = details['buy_price'] * details['quantity']
            curr_price = current_prices.get(stock, details['buy_price'])
            curr_val = curr_price * details['quantity']
            gain = curr_val - buy_val
            print(f"{stock}:{details['quantity']} shares | "
                  f"Bought @ ₹{details['buy_price']} | "
                  f"Now @ ₹{curr_price} | "
                  f"P&L: ₹{gain:.0f}")

        profit, pct = self.pnl(current_prices)
        print(f"\nTotal P&L: ₹{profit:.0f} ({pct:.1f}%)")

rishi = StockPortfolio("Rishi")
rishi.buy("Suzlon", 55, 18)
rishi.buy("IRFC", 150, 6)

today_prices = {"Suzlon": 63, "IRFC": 158}
rishi.show(today_prices)
```
