```python
# Week 1 Build Task — Stock Watchlist Program

my_stocks = ["Suzlon", "IRFC", "CMR Green"]

stock_prices = {
    "Suzlon": 55,
    "IRFC": 150,
    "CMR Green": 192
}

def show_watchlist(stock_list):
    print("=== My Watchlist ===")
    for i, stock in enumerate(stock_list, 1):
        print(f"{i}. {stock}")

def show_prices(prices):
    print("\n=== Current Prices ===")
    for stock, price in prices.items():
        print(f"{stock}: ₹{price}")

def is_in_watchlist(stock_name, watchlist):
    if stock_name in watchlist:
        print(f"{stock_name} is in your watchlist")
    else:
        print(f"{stock_name} is NOT in your watchlist")

show_watchlist(my_stocks)
show_prices(stock_prices)
is_in_watchlist("Suzlon", my_stocks)
is_in_watchlist("TCS", my_stocks)
```
